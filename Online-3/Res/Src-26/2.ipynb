```python
# Chi-squared test
def lcm(a, c, m, x_0):
    # x_0 is seed
    while True:
        x_0 = (a*x_0 + c) & ((1 << 31)-1) # %31 thru bitwise op
        yield x_0 / m # normalizing upto 1


def ctest():
    e = []
    gen = lcm(1103515245,12345,1<<31,27) # gcc values
    l = []
    for i in range(1000):
        l.append(next(gen))
    e, _ = np.histogram(l, bins=np.linspace(0, 1, 11))
    print(l)
    print(e)
    return stats.chisquare(e)

def chi_squared(*args):
    # expected input: init_of_interval, end_of_interval (both inclusive), no_of_bins, the rand values
    n = len(args) - 3
    if n <= 0:
        return -1
    a, b, bins = args[:3]
    print(a, b, bins)
    chi = 0
    for i in range(bins):
        e = n/bins
        o = sum(1 for val in args[3:] if val >= i*(b-a)/bins and val < (i+1)*(b-a)/bins)
        tmp = (((o - e)**2)/e)
        chi += tmp
        print(o, e, tmp)
    print(chi)

gen = lcm(1103515245,12345,1<<31,27) # gcc values
l = []
for i in range(1000):
    l.append(next(gen))
chi_squared(0, 1, 10, *l)

ctest()        
```
