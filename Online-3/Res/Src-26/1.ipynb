```python
import numpy as np
import matplotlib.pyplot as plt
# import scipy as sp
from scipy import stats
```

```python
# rand no gen using linear congruential method
def lcm(a, c, m, x_0):
    # x_0 is seed
    while True:
        x_0 = (a*x_0 + c) & ((1 << 31)-1) # %31 thru bitwise op
        yield x_0 / m # normalizing upto 1

gen = lcm(1103515245,12345,1<<31,27) # gcc values
l = []
for i in range(100):
    l.append(next(gen))

print(l)
```

```python
# stat test
# chisq test
def chi_squared(*args):
    # expected input: init_of_interval, end_of_interval (both inclusive), no_of_bins, the rand values
    n = len(args) - 3
    if n <= 0:
        return -1
    a, b, bins = args[:3]
    print(a, b, bins)
    chi = 0
    for i in range(bins):
        e = n/bins
        o = sum(1 for val in args[3:] if val >= i*(b-a)/bins and val < (i+1)*(b-a)/bins)
        tmp = (((o - e)**2)/e)
        chi += tmp
        print(o, e, tmp)
    print(chi)



chi_squared(0, 1, 10, *l) 
print(stats.chi2.ppf(1-0.05, 9)) # chi square table access
```

```python
def KSTest(*args):
    n = len(args) - 2 # a, b, vals
    a, b = args[:2]
    args = sorted(args[2:])
    args = sorted((v - a)/(b - a) for v in args)

    if n <= 0:
        return -1
    dmax = float('-inf')
    # d+
    for i in range(1, n+1):
        d = (a+(i/n) - args[i-1])
        dmax = max(dmax, d)
    # d-
    for i in range(1, n+1):
        d = (args[i-1] - (a+((i-1)/n)))
        dmax = max(dmax, d)
    print(dmax)

KSTest(0, 1, 0.44,.81,.14,.05,.93)
stats.kstest([0.44,.81,.14,.05,.93], stats.uniform.cdf)
```

```python
# auto-correlation
def autocorrelation(*args):
    #ret z value
    # input: i, l, randvals
    n = len(args) - 2
    ip, l = args[:2]
    args = args[2:]
    M = np.floor((n - ip)/l - 1)
    mean = 0
    print(M)
    for i in range(ip, int(ip+l*(M+1))):
        mean += args[i-1]*args[i+l-1] # -1 due to 0th idxing, slide follows 1th idx
    mean /= (M+1)
    mean -= .25
    var = np.sqrt(13*M+7)/(12*(M+1))
    return mean/var

print(autocorrelation(1,1,0.23,0.28,0.33,0.27,0.05,0.36))
correlation, p_value = stats.pearsonr([0.23,0.28,0.33,0.27,0.05], [0.28,0.33,0.27,0.05,0.36])
print(correlation)
```

```python
# montecarlo integration
def integrate(fn, a, b, samples=10000):
    l = np.zeros(samples)
    gen = lcm(1103515245,12345,1<<31,27)
    for i in range(samples):
        l[i] = next(gen)
    l = a + (b-a) * l

    y = fn(l)
    return (b-a)*sum(y)/len(y) 

    

integrate(np.sin, 0, np.pi)
```

```python
#appx pi using monte carlo
def pi_appx(samples):
    gen = lcm(1103515245,12345,1<<31,27)
    inside_circ = 0
    for i in range(samples):
        x, y = next(gen), next(gen)
        if x**2 + y**2 <= 1:
            inside_circ += 1
    return 4 * inside_circ / (samples)

pi_appx(10000)
        
```
