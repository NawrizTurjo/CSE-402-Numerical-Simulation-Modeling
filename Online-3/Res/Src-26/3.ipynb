```python
#random walk
import numpy as np
import matplotlib.pyplot as plt
# import scipy as sp
from scipy import stats
```

```python
def lcm(a, c, m, x_0, s=0, f=1):
    # x_0 is seed
    while True:
        x_0 = (a*x_0 + c) & ((1 << 31)-1) # %31 thru bitwise op
        yield +1 if s + (x_0 / m) * f > 0.5 else -1   # normalizing upto 1

gen = lcm(1103515245,12345,1<<31,27) # gcc values
count = 0
for i in range(50000):
    l = [next(gen) for i in range(100)]
    k = sum(l)
    if k > 15:
        count += 1

print(count / 50000)


```
